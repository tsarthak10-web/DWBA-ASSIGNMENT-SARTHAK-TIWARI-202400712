```python
!pip install pyspark -q
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum as _sum

spark = SparkSession.builder.appName('SalesAnalysis').getOrCreate()
sales_records = [
    ('ORD101', 'Smartphones', 'Electronics', 15),
    ('ORD102', 'Laptops', 'Electronics', 8),
    ('ORD103', 'Office Chair', 'Furniture', 12),
    ('ORD104', 'Dining Table', 'Furniture', 4),
    ('ORD105', 'T-Shirts', 'Apparel', 35),
    ('ORD106', 'Jeans', 'Apparel', 20),
    ('ORD107', 'Air Conditioners', 'Home Appliances', 6),
    ('ORD108', 'Microwaves', 'Home Appliances', 11)
]
columns = ['OrderID', 'ProductName', 'Category', 'QuantitySold']
df_spark = spark.createDataFrame(sales_records, schema=columns)
df_spark.groupBy('Category').agg(_sum('QuantitySold').alias('TotalQuantitySold')).show()
```
